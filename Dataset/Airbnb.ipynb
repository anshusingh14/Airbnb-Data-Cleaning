```python
# Importing required library
import pandas as pd
import numpy as np
```

```python
# Loading the Airbnb dataset
data = pd.read_csv("Airbnb_Open_Data.csv")

# Displaying first five records
data.head()
```

```python
# Checking the dimensions of the dataset
print("Dataset Shape:", data.shape,"\n")

# Checking column names
print(data.columns,"\n")

# Checking data types and non-null values
data.info()

# Checking statistical summary
data.describe(include="all")
```

```python
# Selecting only the columns required for analysis
columns_to_keep = [
    'NAME', 'host id', 'host_identity_verified', 'host name',
    'neighbourhood group', 'neighbourhood', 'lat', 'long',
    'country', 'country code', 'instant_bookable',
    'cancellation_policy', 'room type', 'Construction year',
    'price', 'service fee', 'minimum nights',
    'number of reviews', 'last review'
]

# Creating a working copy containing selected columns
data = data[columns_to_keep].copy()

# Checking the updated dataset dimensions
print("Shape after column selection:", data.shape)
```

```python
# Renaming Columns for better readability
updated_columns = [column.capitalize() for column in data.columns]
data.columns = updated_columns

# Displaying updated column names
data.columns
```

```python
# Checking the number of duplicate records
duplicate_count = data.duplicated().sum()

print("Duplicate records:", duplicate_count)

# Removing duplicate records
data.drop_duplicates(inplace=True)

# Verifying duplicate removal
print("Duplicates after cleaning:", data.duplicated().sum())
```

```python
# Checking missing values in each column
missing_values = data.isna().sum()

print("Missing values before cleaning:")
print(missing_values,"\n")

# Checking total missing values
print("Total missing values:", data.isna().sum().sum())
```

```python
# Removing rows containing missing values
data.dropna(inplace=True)

# Checking missing values after removal
print("Missing values after cleaning:")
print(data.isna().sum())
```

```python
# Standardising host identity verification values
data["Host_identity_verified"] = (
    data["Host_identity_verified"]
    .str.strip()
    .str.upper()
)

# Converting instant_bookable Boolean values into integers
# True = 1 and False = 0
data["Instant_bookable"] = data["Instant_bookable"].apply(lambda x: 1 if x == True else 0)

# Checking updated values
print(data["Host_identity_verified"].unique())
print(data["Instant_bookable"].unique())
```

```python
# Checking the original price data type
print("Before cleaning:", data["Price"].dtype)

# Removing dollar signs, commas and extra spaces
data["Price"] = (
    data["Price"]
    .astype("string")
    .str.replace(r"[$,\s]", "", regex=True)
)

# Converting price values into numeric format
data["Price"] = pd.to_numeric(data["Price"], errors="coerce")

# Checking the updated data type
print("After cleaning:", data["Price"].dtype)

# Displaying sample price values
data["Price"].head()
```

```python
# Checking final dataset dimensions
print("Final Dataset Shape:", data.shape)

# Checking missing values
print("\nMissing Values:")
print(data.isna().sum())

# Checking duplicate records
print("\nDuplicate Records:", data.duplicated().sum())

# Checking data types
print("\nFinal Data Types:")
print(data.dtypes)

# Displaying first five cleaned records
data.head()
```

```python
# Saving the cleaned dataset as a CSV file
data.to_csv("Cleaned_Airbnb_data.csv", index=False)

print("Cleaned dataset exported successfully!")
```
